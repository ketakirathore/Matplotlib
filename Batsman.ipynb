```python
plt.figure(figsize=(10, 6))  # Create a figure with appropriate size

# For horizontal bar chart, 'Opposition' should be y-axis and 'Runs' should be width
plt.barh(y=df['Opposition'],  # y-coordinates (categories on y-axis)
         width=df['Runs'],    # width of bars (data values)
         height=0.8)          # height of bars (thickness)

plt.xlabel('Runs')
plt.ylabel('Opposition')
plt.title('Runs by Opposition')
plt.tight_layout()  # Adjust layout to fit labels
plt.show()
```
