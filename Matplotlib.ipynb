```python
y =[40,10,20,30]
x = ["A","B","C","D"] 
c = ["k", "purple", "coral","Orange"]
```

```python
plt.pie(y, labels = x, colors = c , startangle = 190, shadow = True, explode = [0.1,0.1,0.1,0.1], counterclock = False, frame= True)
plt.show()
```

```python
x =[3,1,2,5]
y = [5,2,4,7]
```

```python
plt.plot(x,y, 'b*')
plt.axis([0,6,0,20])
for i , j in zip(x,y):
    plt.text(i,j+0.5, '({},{})'.format(i,j))
plt.show()
```
